```python
#Q1]  What are data structures, and why are they important

'''Data structures are ways of organizing and storing data so that it can be accessed and modified efficiently. 
They are important because they help programs run faster, use memory efficiently, and make code easier to understand and manage.'''
```

```python
 # Q2 ]  Explain the difference between mutable and immutable data types with examples.

'''Mutable data types can be changed after creation, while immutable data types cannot.

Mutable example: List
Immutable example: Tuple'''
```

```python
# Q3] What are the main differences between lists and tuples in Python3

'''Lists are mutable, while tuples are immutable. Lists use more memory and are slower compared to tuples. Tuples are safer for storing fixed data.'''
```

```python
# Q4]  Describe how dictionaries store data.

'''Dictionaries store data in key–value pairs. Each key is unique and mapped to a value using a hashing technique, which allows fast data access.'''
```

```python
# Q5] Why might you use a set instead of a list in Python.

'''Sets automatically remove duplicate values and provide faster membership testing compared to lists.'''
```

```python
# Q6]  What is a string in Python, and how is it different from a list.

'''A string is a sequence of characters, while a list is a collection of elements. Strings are immutable, but lists are mutable.'''
```

```python
# Q7] How do tuples ensure data integrity in Python3

'''Since tuples cannot be modified, they protect data from accidental changes, ensuring consistency.'''
```

```python
#Q8]  What is a hash table, and how does it relate to dictionaries in Python.

'''A hash table stores data using key-value mapping. Python dictionaries internally use hash tables for fast data retrieval.'''
```

```python
# Q9] Can lists contain different data types in Python

'''Yes, lists in Python can store elements of different data types.'''
```

```python
# Q10] Explain why strings are immutable in Python

'''Strings are immutable to improve performance and memory efficiency and to maintain data safety.'''
```

```python
# Q11] What advantages do dictionaries offer over lists for certain tasks

'''Dictionaries allow faster lookups using keys, whereas lists require searching through elements.'''
```

```python
# Q12] Describe a scenario where using a tuple would be preferable over a listP

'''Tuples are preferred when data should not change, such as storing fixed configuration values.'''
```

```python
# Q13] How do sets handle duplicate values in Python.

'''Sets automatically remove duplicate values and store only unique elements.'''
```

```python
# Q14] How does the “in” keyword work differently for lists and dictionaries3

'''In lists, in checks for values
   In dictionaries, in checks for keys'''
```

```python
# Q15] Can you modify the elements of a tuple? Explain why or why notP

'''No, tuple elements cannot be modified because tuples are immutable.'''
```

```python
# Q16] What is a nested dictionary, and give an example of its use case.

'''A nested dictionary is a dictionary inside another dictionary.'''
```

```python
# Q17] Describe the time complexity of accessing elements in a dictionary.

'''Accessing elements in a dictionary has O(1) average time complexity.'''
```

```python
# Q18]  In what situations are lists preferred over dictionaries3

'''Lists are preferred when order matters and data is accessed by position.'''
```

```python
# Q19] Why are dictionaries considered unordered, and how does that affect data retrieval

'''Dictionaries do not store items in a fixed index order; data is retrieved using keys.'''
```

```python
# 20]  Explain the difference between a list and a dictionary in terms of data retrieval.

'''Lists retrieve data using index positions, while dictionaries retrieve data using keys.'''
```
