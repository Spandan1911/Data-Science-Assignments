```python
 #  Q1]  What is Python, and why is it popular 

'''Python is a high-level, interpreted programming language that is easy to read and write. It uses simple English-like syntax, which makes it 
beginner-friendly. Python is popular because it supports multiple programming styles, has a large collection of libraries, and is widely 
used in areas like web development, data science, artificial intelligence, automation, and software testing.'''
```

```python
# Q2] What is an interpreter in Python 

'''An interpreter is a program that reads and executes Python code line by line. Python does not need compilation before execution. 
This makes debugging easier because errors are shown immediately at the line where they occur.'''
```

```python
# Q3] What are pre-defined keywords in Python 

'''Pre-defined keywords are reserved words in Python that have a special meaning. They are part of Python’s syntax and cannot be used for 
other purposes. Examples include if, else, for, while, True, False, and None.'''
```

```python
# Q4] Can keywords be used as variable names 

'''No, Python keywords cannot be used as variable names. If we try to do so, Python will raise a syntax error because keywords are reserved for 
specific operations in the language.'''
```

```python
# Q5] What is mutability in Python 

'''Mutability refers to whether an object’s value can be changed after it is created. If an object can be modified, it is mutable; if it cannot
be changed, it is immutable.'''
```

```python
# Q6] Why are lists mutable, but tuples are immutable 

'''Lists are designed to be flexible, so elements can be added, removed, or changed. Tuples are immutable to ensure data safety
and consistency. This makes tuples faster and useful when data should not be altered.'''
```

```python
# Q7] What is the difference between “==” and “is” operators in Python 

'''== checks whether two variables have the same value.
   is checks whether two variables refer to the same memory location.'''
```

```python
# Q8] What are logical operators in Python 

'''Logical operators are used to combine conditions:

        and – returns True if both conditions are True
        or – returns True if at least one condition is True
        not – reverses the result'''
```

```python
# Q9] What is type casting in Python 

'''Type casting is the process of converting one data type into another, such as converting a string to an integer or a float.'''
```

```python
# Q10] What is the difference between implicit and explicit type casting 

'''Implicit type casting happens automatically (e.g., int to float).
   Explicit type casting is done manually using functions like int(), float(), and str().'''
```

```python
# Q11] What is the purpose of conditional statements in Python 

'''Conditional statements allow a program to make decisions based on conditions. They control the flow of execution using statements 
  like if, elif, and else.'''
```

```python
# Q12] How does the elif statement work 

'''The elif statement checks multiple conditions. If the first if condition is false, Python checks the elif condition before moving to else.'''
```

```python
# Q13]  What is the difference between for and while loops 

'''for loop is used when the number of iterations is known.
   while loop is used when the condition depends on logic rather than count.'''
```

```python
# Q14] Describe a scenario where a while loop is more suitable than a for loop

'''A while loop is suitable when the number of iterations is unknown, such as repeatedly asking a user for input until a correct value is entered.'''
```

```python
# Q5] Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each.

def modify_list(lst):
    lst.append(10)

def modify_number(num):
    num += 10

my_list = [1, 2, 3]
my_num = 5

modify_list(my_list)
modify_number(my_num)

print(my_list)
print(my_num)
```

```python
# Q11] Write a program that checks if a number is positive, negative, or zero.

num = int(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
# Q14] Write a program to reverse a string using a while loop.

s = input("Enter a string: ")
i = len(s) - 1

while i >= 0:
    print(s[i], end="")
    i -= 1
```

```python
# Q15] Write a Python program to calculate the factorial of a number provided by the user using a while loop.

num = int(input("Enter a number: "))
fact = 1

while num > 0:
    fact *= num
    num -= 1

print("Factorial:", fact)
```
